```python
from sentence_transformers import SentenceTransformer
from PIL import Image
from IPython.display import display

# Load CLIP model
model = SentenceTransformer("clip-ViT-B-32")

# Encode an image:
img = Image.open("images/small/00/00a69c51.jpg")
img_emb = model.encode(img)

# Encode text descriptions
texts = ["Two dogs in the snow", "Kitchen utilities", "A picture of London at night", "shop"]
text_emb = model.encode(
    texts
)

# Display the image
display(img)

# Compute similarities
similarity_scores = model.similarity(img_emb, text_emb)[0]
for text, score in zip(texts, similarity_scores):
    print(f"Similarity with '{text}': {score:.4f}")
```
