```python
META_FILENAME = "images/metadata/images.csv"
IMAGES_DIR = "images/small"
```

```python
import pandas as pd

def load_metadata():
    metadata = pd.read_csv(META_FILENAME)
    return metadata

meta = load_metadata()
meta.head()
```

```python
# find small images
def find_small_images(df):
    df = df[(df['width'] < 1000) | (df['height'] < 1000)]
    return df

smol = find_small_images(meta)
print(f"Small images to be removed ({len(smol)} of {len(meta)}):")
smol.head()
```

```python
import os
def remove_images(df: pd.DataFrame, images_dir:str, dry_run:bool=False) -> int:
    total_removed = 0
    for _, row in df.iterrows():
        image_path = f"{images_dir}/{row['path']}"
        try:
            if not dry_run:
                os.remove(image_path)
            total_removed += 1
        except FileNotFoundError:
            pass
    return total_removed
        
tot = remove_images(smol, IMAGES_DIR, dry_run=False)
print(f"Total images removed: {tot}")
```
